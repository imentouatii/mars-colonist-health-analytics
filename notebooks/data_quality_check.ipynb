```python
import os
import pandas as pd
import numpy as np
from IPython.display import display
```

```python
DATA_DIR = r"H:\Downloads\Move to Mars\raw_data"
```

```python
def check_data_quality(df, name):
    print(f"\n==================================================")
    print(f" ANALYSE DE LA TABLE : {name}")
    print(f"==================================================")
    print(f"• Dimensions : {df.shape[0]} lignes, {df.shape[1]} colonnes")
    
    # 1. Doublons stricts
    dup_count = df.duplicated().sum()
    print(f"• Doublons stricts : {dup_count} ({dup_count/len(df)*100:.2f}%)")
    
    # 2. Valeurs manquantes
    missing = df.isnull().sum()
    missing_cols = missing[missing > 0]
    if not missing_cols.empty:
        print("• Valeurs manquantes :")
        for col, count in missing_cols.items():
            print(f"  - {col}: {count} NULL ({count/len(df)*100:.2f}%)")
    else:
        print("• Valeurs manquantes : Aucune (100% complet)")
        
    # 3. Métadonnées des colonnes (Noms, Types & Exemples)
    print("\n Détail des colonnes et types de données :")
    col_summary = pd.DataFrame({
        'Type': df.dtypes.astype(str),
        'Non-Null': df.notnull().sum(),
        'Valeurs Uniques': df.nunique(),
        'Exemple': [df[col].iloc[0] if len(df) > 0 else None for col in df.columns]
    })
    display(col_summary)
    
    # 4. Aperçu des 3 premières lignes (Affiche le tableau HTML propre)
    print("\n Aperçu des 3 premières lignes :")
    display(df.head(3))
```

```python
# ----------------------------------------------------
# 1. TABLE : COLON
# ----------------------------------------------------
df_colon = pd.read_csv(os.path.join(DATA_DIR, "COLON.csv"))
check_data_quality(df_colon, "COLON")
# Vérification des Unicités / Clé Primaire
print(" -> Unicité id_colon :", df_colon['id_colon'].is_unique)
```

```python
# TABLE : CONSTANTE_VITALE (Table de faits)
df_constantes = pd.read_csv(os.path.join(DATA_DIR, "CONSTANTE_VITALE.csv"))
check_data_quality(df_constantes, "CONSTANTE_VITALE")

# Vérification des bornes physiologiques (Outliers)
print("\n• Outliers Physiologiques dans CONSTANTE_VITALE :")
for col, (vmin, vmax) in {'spo2': (70, 100), 'bpm': (40, 180), 'temperature': (35, 42)}.items():
    if col in df_constantes.columns:
        outliers = df_constantes[(df_constantes[col] < vmin) | (df_constantes[col] > vmax)]
        print(f"  - Anomalies [{col}] (hors {vmin}-{vmax}) : {len(outliers)} lignes")

```

```python
# TABLE : CATEGORIE_METIER
df_metier = pd.read_csv(os.path.join(DATA_DIR, "CATEGORIE_METIER.csv"))
check_data_quality(df_metier, "CATEGORIE_METIER")
```
